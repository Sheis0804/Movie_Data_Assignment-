### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
movies['No_of_Votes'] = pd.to_numeric(movies['No_of_Votes'], errors='coerce')
movies_clean = movies.dropna(subset=['No_of_Votes'])
top = movies_clean.nlargest(5, 'No_of_Votes')
plt.figure(figsize=(10, 5))
plt.barh(top['Series_Title'], top['No_of_Votes'])
plt.gca().invert_yaxis()
plt.xlabel('Number of Votes')
plt.ylabel('Movie Title')
plt.title('Top 5 Movies by Number of Votes')
plt.tight_layout()
plt.show()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
d = movies[movies['Genre'].str.contains('Drama|Action', regex=True, na=False)]
drama = d[d['Genre'].str.contains('Drama', na=False)]['Gross'].dropna()

action = d[d['Genre'].str.contains('Action', na=False)]['Gross'].dropna()
plt.figure(figsize=(7, 5))
plt.boxplot([drama, action], tick_labels=['Drama', 'Action'])
plt.ylabel('Gross Revenue')
plt.title('Gross Revenue: Drama vs Action Movies')
plt.show()
```
